```python
import os
from histolab.slide import Slide
from histolab.tiler import GridTiler

# 1. Ruta de tu archivo .mrxs
ruta_mrxs = "C:/Users/Abrah/OneDrive/Documents/Patologia Computacional/sample/01_01_Histological_images/43.mrxs"

# 2. Cargar el slide con histolab
slide = Slide(ruta_mrxs, processed_path="dataset_histolab/")

# 3. Configurar el extractor de parches (GridTiler)
# tissue_percent=90 significa: "Solo guarda el parche si el 90% de sus píxeles son tejido real (no fondo blanco)"
grid_tiler = GridTiler(
    tile_size=(256, 256),
    level=2,             # El mismo nivel que usaste antes
    check_tissue=True,   # Activar el filtro automático de tejido
    tissue_percent=90.0, # Un umbral alto para descartar mucha grasa/fondo
    pixel_overlap=0
)

# 4. Extraer y guardar automáticamente
print("Extrayendo parches...")
grid_tiler.extract(slide)
print("¡Parches extraídos y guardados en dataset_histolab/tiles/!")
```
